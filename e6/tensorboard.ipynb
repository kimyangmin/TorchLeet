## PyTorch에서 TensorBoard로 학습 진행 상황 시각화하기

### 문제 설명
PyTorch의 선형 회귀 모델 학습 진행 상황을 모니터링하기 위해 **TensorBoard**를 사용하는 과제입니다. TensorBoard는 학습 중 손실 같은 지표를 추적할 수 있는 시각적 인터페이스를 제공하므로, 모델을 분석하고 디버깅하는 데 도움이 됩니다.

### 요구사항
1. **TensorBoard 통합**:
   - `torch.utils.tensorboard.SummaryWriter`를 사용해 TensorBoard를 설정합니다.
   - 각 에포크가 끝날 때 학습 손실을 기록합니다.

4. **시각화**:
   - 다음 명령으로 TensorBoard를 시작합니다:
     ```bash
     tensorboard --logdir=runs
     ```
   - 학습 중 손실 곡선을 시각화합니다.

### 제약 사항
- TensorBoard 로그는 `runs`라는 이름의 디렉터리에 저장되어야 합니다.
- 이 해법은 여러 에포크를 처리하고 손실을 일관되게 기록해야 합니다.


````python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.tensorboard import SummaryWriter
````

````python
# Generate synthetic data
torch.manual_seed(42)
X = torch.rand(100, 1) * 10  # 100 data points between 0 and 10
y = 3 * X + 5 + torch.randn(100, 1)  # Linear relationship with noise

# Define a simple Linear Regression Model
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super(LinearRegressionModel, self).__init__()
        self.linear = nn.Linear(1, 1)  # Single input and single output

    def forward(self, x):
        return self.linear(x)
````

````python
# TODO: Initialize SummaryWriter to log data
# TODO: 스터디 레퍼런스:
# https://pytorch.org/tutorials/recipes/recipes/tensorboard_with_pytorch.html

writer = SummaryWriter()

# Initialize the model, loss function, and optimizer
model = LinearRegressionModel()
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

# Training loop
epochs = 100
for epoch in range(epochs):
    # Forward pass
    predictions = model(X)
    loss = criterion(predictions, y)

    # Backward pass and optimization
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    # Log loss to TensorBoard
    # TODO: Add the loss value to TensorBoard
    writer.add_scalar('Loss/train', loss.item(), epoch) # 스칼라 loss 값을 epoch랑 함께 기록

    # Log progress every 10 epochs
    if (epoch + 1) % 10 == 0:
        print(f"Epoch [{epoch + 1}/{epochs}], Loss: {loss.item():.4f}")

# Close the TensorBoard writer
writer.close()

# Run TensorBoard using the logs generated
# Command to run: tensorboard --logdir=runs
````

`tensorboard --logdir=./runs`을 실행시키면 됩니다.
