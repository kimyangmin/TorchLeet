# 심층 신경망 작성하기

### 문제 설명
PyTorch를 사용해 비선형 관계를 보이는 합성 데이터에서 회귀 문제를 해결하기 위한 **심층 신경망(DNN)** 모델을 구성합니다. 목표는 합성 데이터에서 목표 값을 예측하는 것입니다.

### 요구사항
`DNNModel` 클래스를 다음 기준에 맞게 구현합니다.

1. **모델 정의**:
   - 모델에는 다음이 포함되어야 합니다:
     - **입력 레이어**와 연결된 **은닉 레이어**.
     - 비선형성을 위한 **ReLU 활성화 함수**.
     - 회귀를 위한 단일 유닛을 가진 **출력 레이어**.

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# Generate synthetic data
torch.manual_seed(42)
X = torch.rand(100, 2) * 10  # 100 data points with 2 features
y = (X[:, 0] + X[:, 1] * 2).unsqueeze(1) + torch.randn(100, 1)  # Non-linear relationship with noise
```

```python
# Define the Deep Neural Network Model
class DNNModel(nn.Module):
    def __init__(self, input_size = 2, hidden_size = 3, output_size = 1):
        super(DNNModel, self).__init__()
        # TODO: Explain about nn.Sequential and when is it used.
        # Cite it and update it on my notes.
        # https://pytorch.org/docs/stable/generated/torch.nn.Sequential.html
        self.model = nn.Sequential(
            nn.Linear(input_size, hidden_size), # up_proj
            nn.ReLU(), # activation
            nn.Linear(hidden_size, output_size), # down_proj to target
            # No activation for output,
        )

    def forward(self, x):
        x = self.model(x)
        return x
```

```python
# Initialize the model, loss function, and optimizer
model = DNNModel()
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

# Training loop
epochs = 1000
for epoch in range(epochs):
    # Forward pass
    predictions = model(X)
    loss = criterion(predictions, y)

    # Backward pass and optimization
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    # Log progress every 100 epochs
    if (epoch + 1) % 100 == 0:
        print(f"Epoch [{epoch + 1}/{epochs}], Loss: {loss.item():.4f}")

# Testing on new data
X_test = torch.tensor([[4.0, 3.0], [7.0, 8.0]])
with torch.no_grad():
    predictions = model(X_test)
    print(f"Predictions for {X_test.tolist()}: {predictions.tolist()}")
```
