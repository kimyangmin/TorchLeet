## GAN 작성하기

### 문제 설명
필요한 섹션을 완성하여 **생성적 적대 신경망(GAN)** 을 구현합니다. GAN은 가짜 데이터를 생성하는 **Generator**와, 데이터를 진짜인지 가짜인지 분류하는 **Discriminator**로 구성됩니다.

### 요구사항
1. **Generator 클래스 정의**:
   - **목적**: 실제 데이터 분포를 모방하는 가짜 데이터를 생성합니다.
   - **레이어**:
     - 잠재 공간(무작위 노이즈)을 더 높은 차원의 공간으로 매핑하는 완전 연결 레이어로 시작합니다.
     - 비선형성을 추가하기 위해 `ReLU` 같은 활성화 함수를 사용합니다.
     - 데이터를 처리하고 구조를 다듬기 위해 추가 레이어를 더합니다.
     - 마지막 레이어는 목표 형태의 데이터를 출력해야 합니다. 스케일링을 위해 `Tanh` 같은 활성화 함수를 사용합니다.
   - **Forward Pass**: 정의한 레이어를 통과하도록 forward 메서드를 구현합니다.

2. **Discriminator 클래스 정의**:
   - **목적**: 데이터를 진짜인지 가짜인지 분류합니다.
   - **레이어**:
     - 입력을 처리하고 특징을 추출하기 위해 완전 연결 레이어를 사용합니다.
     - 죽은 뉴런을 방지하고 학습을 안정화하기 위해 `LeakyReLU` 같은 활성화 함수를 적용합니다.
     - 마지막 레이어는 `Sigmoid` 활성화를 사용해 단일 확률(진짜 또는 가짜)을 출력해야 합니다.
   - **Forward Pass**: 레이어를 통과하도록 forward 메서드를 구현합니다.

3. **GAN 학습**:
   - Generator와 Discriminator의 학습을 번갈아 수행합니다.
   - 두 모델 모두에 이진 교차 엔트로피 손실을 사용합니다.
   - 학습 중 손실과 생성 샘플을 모니터링합니다.

```python
# Generate synthetic data for training
torch.manual_seed(42)
real_data = torch.rand(100, 1) * 2 - 1  # 100 samples in the range [-1, 1]

# Initialize models, loss, and optimizers
latent_dim = 10
data_dim = 1
G = Generator(latent_dim, data_dim)
D = Discriminator(data_dim)

criterion = nn.BCELoss()
optimizer_G = optim.Adam(G.parameters(), lr=0.001)
optimizer_D = optim.Adam(D.parameters(), lr=0.001)
```

```python
# Training loop
epochs = 1000
for epoch in range(epochs):
    # Train Discriminator
    latent_samples = torch.randn(real_data.size(0), latent_dim)
    fake_data = G(latent_samples).detach()
    real_labels = torch.ones(real_data.size(0), 1)
    fake_labels = torch.zeros(real_data.size(0), 1)

    optimizer_D.zero_grad()
    real_loss = criterion(D(real_data), real_labels)
    fake_loss = criterion(D(fake_data), fake_labels)
    loss_D = real_loss + fake_loss
    loss_D.backward()
    optimizer_D.step()

    # Train Generator
    latent_samples = torch.randn(real_data.size(0), latent_dim)
    fake_data = G(latent_samples)
    optimizer_G.zero_grad()
    loss_G = criterion(D(fake_data), real_labels)
    loss_G.backward()
    optimizer_G.step()

    # Log progress every 100 epochs
    if (epoch + 1) % 100 == 0:
        print(f"Epoch [{epoch + 1}/{epochs}] - Loss D: {loss_D.item():.4f}, Loss G: {loss_G.item():.4f}")
```

```python
# Generate new samples with the trained Generator
latent_samples = torch.randn(5, latent_dim)
with torch.no_grad():
    generated_data = G(latent_samples)
    print(f"Generated data: {generated_data.tolist()}")
```
